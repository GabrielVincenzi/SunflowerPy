```python
import json
```

## Control for missing naming

```python
with open('library/datasets.json', 'r') as file:
    data = json.load(file)

with open('library/dictionaries.json', 'r') as file:
    names = json.load(file)

names = {k.lower(): v for k, v in names.items()}
names_set = set(names)
values_set = set()

for db in data:
    for variable in data[db]:
        for char in data[db][variable]:
            if char != "description":
                values_set.update(x.strip().lower() for x in data[db][variable][char].split(","))

missing = values_set - names_set
print("There are " + str(len(missing)) + " missing values:")
for value in missing:
    print(value)
```
